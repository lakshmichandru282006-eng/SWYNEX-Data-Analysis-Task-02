```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("Cleaned_Dataset_Task_01.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("DATASET INFORMATION")
print("-------------------")

df.info()

print("\nCOLUMN NAMES")
print("------------")
print(df.columns.tolist())

print("\nDATA TYPES")
print("----------")
print(df.dtypes)
```

```python
print("MISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE RECORDS")
print(df.duplicated().sum())

print("\nUNIQUE VALUES IN CATEGORICAL COLUMNS")

for col in df.select_dtypes(include="object").columns:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

```python
print("DESCRIPTIVE STATISTICAL ANALYSIS")
print("=" * 50)

numerical_columns = [
    'age',
    'scholarship',
    'hipertension',
    'diabetes',
    'alcoholism',
    'handcap',
    'sms_received'
]

print("\nSTATISTICAL SUMMARY")
display(df[numerical_columns].describe())

print("\nMEAN VALUES")
display(df[numerical_columns].mean())

print("\nMEDIAN VALUES")
display(df[numerical_columns].median())

print("\nMODE VALUES")
display(df[numerical_columns].mode().iloc[0])

print("\nSTANDARD DEVIATION")
display(df[numerical_columns].std())
```

```python
print("PATIENT DEMOGRAPHIC ANALYSIS")
print("=" * 50)

print("\nGENDER DISTRIBUTION")
display(df['gender'].value_counts())

print("\nGENDER PERCENTAGE")
display(df['gender'].value_counts(normalize=True) * 100)

print("\nAGE GROUP DISTRIBUTION")

df['age_group'] = pd.cut(
    df['age'],
    bins=[-1, 12, 18, 30, 45, 60, 75, 120],
    labels=[
        'Children',
        'Teenagers',
        'Young Adults',
        'Adults',
        'Middle Aged',
        'Seniors',
        'Elderly'
    ]
)

display(df['age_group'].value_counts().sort_index())
```

```python
plt.figure(figsize=(7, 5))

df['gender'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Gender Distribution of Patients')
plt.ylabel('')
plt.show()


plt.figure(figsize=(9, 5))

sns.histplot(
    df['age'],
    bins=30,
    kde=True
)

plt.title('Age Distribution of Patients')
plt.xlabel('Age')
plt.ylabel('Number of Appointments')
plt.show()
```

```python
print("APPOINTMENT ATTENDANCE ANALYSIS")
print("=" * 45)

attendance = df['no-show'].value_counts()

print("\nAttendance Count:")
display(attendance)

print("\nAttendance Percentage:")
display(df['no-show'].value_counts(normalize=True) * 100)

total = len(df)

attended = (df['no-show'] == 'No').sum()
missed = (df['no-show'] == 'Yes').sum()

print("\nTotal Appointments:", total)
print("Attended Appointments:", attended)
print("Missed Appointments:", missed)

print("Attendance Rate:", round(attended / total * 100, 2), "%")
print("No-Show Rate:", round(missed / total * 100, 2), "%")
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x='no-show',
    hue='no-show',
    palette='Set2',
    legend=False
)

plt.title('Patient Appointment Attendance')
plt.xlabel('No-Show Status')
plt.ylabel('Number of Appointments')
plt.show()
```

```python
medical_columns = [
    'hipertension',
    'diabetes',
    'alcoholism',
    'handcap'
]

medical_counts = df[medical_columns].sum()

print("MEDICAL CONDITION DISTRIBUTION")
display(medical_counts)

plt.figure(figsize=(9, 5))

medical_counts.sort_values(ascending=False).plot(
    kind='bar',
    edgecolor='black'
)

plt.title('Distribution of Medical Conditions')
plt.xlabel('Medical Conditions')
plt.ylabel('Number of Patients')
plt.xticks(rotation=30)
plt.show()
```

```python
sms_analysis = pd.crosstab(
    df['sms_received'],
    df['no-show'],
    normalize='index'
) * 100

print("SMS VS APPOINTMENT ATTENDANCE")
display(sms_analysis)

sms_analysis.plot(
    kind='bar',
    stacked=True,
    figsize=(8, 5),
    colormap='Set2'
)

plt.title('SMS Notifications vs Appointment Attendance')
plt.xlabel('SMS Received (0 = No, 1 = Yes)')
plt.ylabel('Percentage')
plt.legend(title='No-Show Status')
plt.xticks(rotation=0)
plt.show()
```

```python
df['scheduledday'] = pd.to_datetime(df['scheduledday'])
df['appointmentday'] = pd.to_datetime(df['appointmentday'])

df['waiting_days'] = (
    df['appointmentday'].dt.normalize() -
    df['scheduledday'].dt.normalize()
).dt.days

print("WAITING TIME STATISTICS")

display(df['waiting_days'].describe())

plt.figure(figsize=(9, 5))

sns.histplot(
    df['waiting_days'],
    bins=30,
    kde=True
)

plt.title('Distribution of Appointment Waiting Time')
plt.xlabel('Waiting Days')
plt.ylabel('Number of Appointments')
plt.show()
```

```python
print("NEIGHBOURHOOD ANALYSIS")
print("=" * 45)

neighbourhood_counts = df['neighbourhood'].value_counts()

print("\nTOP 10 NEIGHBOURHOODS BY APPOINTMENTS")
display(neighbourhood_counts.head(10))

neighbourhood_no_show = (
    df.groupby('neighbourhood')['no-show']
    .apply(lambda x: (x == 'Yes').mean() * 100)
    .sort_values(ascending=False)
)

print("\nTOP 10 NEIGHBOURHOODS BY NO-SHOW RATE")
display(neighbourhood_no_show.head(10))

plt.figure(figsize=(12, 6))

neighbourhood_counts.head(10).sort_values().plot(kind='barh')

plt.title('Top 10 Neighbourhoods by Appointments')
plt.xlabel('Number of Appointments')
plt.ylabel('Neighbourhood')
plt.show()
```

```python
age_attendance = pd.crosstab(
    df['age_group'],
    df['no-show'],
    normalize='index'
) * 100

print("AGE GROUP VS ATTENDANCE")
display(age_attendance)

age_attendance.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6),
    colormap='Set2'
)

plt.title('Appointment Attendance Across Age Groups')
plt.xlabel('Age Group')
plt.ylabel('Percentage')
plt.legend(title='No-Show Status')
plt.xticks(rotation=30)
plt.show()
```

```python
correlation_columns = [
    'age',
    'scholarship',
    'hipertension',
    'diabetes',
    'alcoholism',
    'handcap',
    'sms_received',
    'waiting_days'
]

correlation = df[correlation_columns].corr()

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Patient Data')
plt.show()
```

```python
print("ANOMALY DETECTION")
print("=" * 45)

print("\nNEGATIVE AGE RECORDS:")
display(df[df['age'] < 0])

print("\nAGE ABOVE 100:")
display(df[df['age'] > 100][['age', 'gender']].head(20))

print("\nNEGATIVE WAITING DAYS:")
display(df[df['waiting_days'] < 0][
    ['scheduledday', 'appointmentday', 'waiting_days']
].head(20))

print("\nWAITING TIME OUTLIERS")

Q1 = df['waiting_days'].quantile(0.25)
Q3 = df['waiting_days'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['waiting_days'] < lower_bound) |
    (df['waiting_days'] > upper_bound)
]

print("Number of waiting time outliers:", len(outliers))
```

```python
print("KEY BUSINESS INSIGHTS")
print("=" * 50)

print("\n1. OVERALL NO-SHOW RATE")

print(round((df['no-show'] == 'Yes').mean() * 100, 2), "%")

print("\n2. GENDER WITH HIGHER NO-SHOW RATE")

gender_rate = df.groupby('gender')['no-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

display(gender_rate)

print("\n3. AGE GROUP WITH HIGHEST NO-SHOW RATE")

age_rate = df.groupby('age_group', observed=True)['no-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

display(age_rate.sort_values(ascending=False))

print("\n4. SMS VS NO-SHOW RATE")

sms_rate = df.groupby('sms_received')['no-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

display(sms_rate)

print("\n5. AVERAGE WAITING DAYS BY ATTENDANCE")

waiting_analysis = df.groupby('no-show')['waiting_days'].mean()

display(waiting_analysis)

print("\n6. TOP NEIGHBOURHOOD BY NO-SHOW RATE")

display(neighbourhood_no_show.head(5))
```

```python
# TEMPORAL TREND ANALYSIS

df['appointment_date'] = pd.to_datetime(
    df['appointmentday']
).dt.normalize()

df['appointment_month'] = df['appointment_date'].dt.month_name()

df['appointment_weekday'] = df['appointment_date'].dt.day_name()

monthly_trend = df.groupby('appointment_month').agg(
    Total_Appointments=('appointmentid', 'count'),
    No_Show_Rate=('no-show', lambda x: (x == 'Yes').mean() * 100)
)

weekday_trend = df.groupby('appointment_weekday').agg(
    Total_Appointments=('appointmentid', 'count'),
    No_Show_Rate=('no-show', lambda x: (x == 'Yes').mean() * 100)
)

print("MONTHLY APPOINTMENT TRENDS")
display(monthly_trend)

print("WEEKDAY APPOINTMENT TRENDS")
display(weekday_trend)

# MONTHLY APPOINTMENT CHART

monthly_trend['Total_Appointments'].plot(
    kind='line',
    marker='o',
    figsize=(10, 5),
    title='Monthly Appointment Trends'
)

plt.ylabel('Total Appointments')
plt.xlabel('Month')
plt.grid(True)
plt.show()

# WEEKDAY NO-SHOW PATTERNS

weekday_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

weekday_trend = weekday_trend.reindex(weekday_order)

weekday_trend['No_Show_Rate'].plot(
    kind='bar',
    figsize=(10, 5),
    title='No-Show Rate by Weekday'
)

plt.ylabel('No-Show Rate (%)')
plt.xlabel('Weekday')
plt.xticks(rotation=30)
plt.show()


# WAITING TIME VS ATTENDANCE PATTERN

print("WAITING TIME VS ATTENDANCE")
display(df.groupby('no-show')['waiting_days'].agg(
    ['mean', 'median', 'min', 'max']
))


# ANOMALY SUMMARY

print("ANOMALY DETECTION")
print("Negative ages:", (df['age'] < 0).sum())
print("Ages above 100:", (df['age'] > 100).sum())
print("Negative waiting days:", (df['waiting_days'] < 0).sum())
print("Waiting time outliers:", len(outliers))

print("Duplicate records:", df.duplicated().sum())
print("Missing values:", df.isnull().sum().sum())
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

os.makedirs("Visualizations", exist_ok=True)

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(4, 2, figsize=(16, 24))

fig.suptitle(
    "MEDICAL APPOINTMENT EXPLORATORY DATA ANALYSIS",
    fontsize=20,
    fontweight="bold",
    y=0.995
)

# 1. Gender Distribution
df['gender'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90,
    ax=axes[0, 0]
)
axes[0, 0].set_title("Gender Distribution")
axes[0, 0].set_ylabel("")

# 2. Age Distribution
sns.histplot(
    data=df,
    x='age',
    bins=30,
    kde=True,
    ax=axes[0, 1]
)
axes[0, 1].set_title("Age Distribution")
axes[0, 1].set_xlabel("Patient Age")

# 3. Appointment Attendance
sns.countplot(
    data=df,
    x='no-show',
    hue='no-show',
    palette='Set2',
    legend=False,
    ax=axes[1, 0]
)
axes[1, 0].set_title("Appointment Attendance")
axes[1, 0].set_xlabel("No-Show Status")

# 4. Medical Conditions
df[['hipertension', 'diabetes', 'alcoholism', 'handcap']].sum().plot(
    kind='bar',
    ax=axes[1, 1]
)
axes[1, 1].set_title("Medical Condition Distribution")
axes[1, 1].set_xlabel("Medical Conditions")
axes[1, 1].tick_params(axis='x', rotation=15)

# 5. SMS Analysis
sms_analysis.plot(
    kind='bar',
    stacked=True,
    colormap='Set2',
    ax=axes[2, 0]
)
axes[2, 0].set_title("SMS vs Appointment Attendance")
axes[2, 0].set_xlabel("SMS Received")
axes[2, 0].set_ylabel("Percentage")
axes[2, 0].tick_params(axis='x', rotation=0)

# 6. Waiting Time
sns.histplot(
    data=df,
    x='waiting_days',
    bins=30,
    kde=True,
    ax=axes[2, 1]
)
axes[2, 1].set_title("Appointment Waiting Time")
axes[2, 1].set_xlabel("Waiting Days")

# 7. Age Group Analysis
age_attendance.plot(
    kind='bar',
    stacked=True,
    colormap='Set2',
    ax=axes[3, 0]
)
axes[3, 0].set_title("Age Group vs Attendance")
axes[3, 0].set_ylabel("Percentage")
axes[3, 0].tick_params(axis='x', rotation=20)

# 8. Correlation Heatmap
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    ax=axes[3, 1]
)
axes[3, 1].set_title("Correlation Heatmap")

plt.tight_layout(rect=[0, 0, 1, 0.98])

fig.savefig(
    "Visualizations/EDA_Complete_Dashboard.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
plt.close(fig)

print("All 8 graphs arranged successfully!")
```
